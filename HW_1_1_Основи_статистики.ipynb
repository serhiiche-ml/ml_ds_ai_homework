`1 3 4`
Виведіть результат кожного обчислення на екран.

```python
x = [1, 3, 4]

sum_x = sum(x)
power_x = sum([el ** 2 for el in x])
sum_x_minus_1_square = sum((el - 1) ** 2 for el in x)

display(Math("\\Sigma x = %s" % (sum_x)))
display(Math("\\Sigma x^2 = %s" % (power_x)))
display(Math("\\Sigma (x-1)^2 = %s" % (sum_x_minus_1_square)))
```

**Завдання 4**. Обчислість стандартне відхилення наведеної вибірки в змінній `sum_purchases` кожним з зазначених методів:  

  4.1. без використання жодних бібліотек, користуючись тільки Python.

**Завдання 6**. Крім даних про суму покупок, ми зібрали ще дані про кількість відвідувань кожного з кожного покупця, які знаходяться в змінній `visits`. Обчисліть коефіцієнт кореляції Пірсона для суми покупок і кількості відвідувань кожним з зазначених методів:  

  6.1. без використання жодних бібліотек, користуючись тільки Python.  
  6.2. використовуючи відповідний метод бібліотеки numpy.  
  
Виведіть результат на екран. Проаналізуйте результат.

Виведіть додатково графік розсіювання викликавши код нижче і зробіть заключення, чи варто щось змінити при підрахунку кореляції, аби коефіцієнт Пірсона показував коректну оцінку лінійної залежності?

```python
visits = [ 6,  2, 10,  3,  9,  9,  2,  3,  6,  6, 10,  2,  5, 10,  3,  5,  4,
        7,  4,  2]
```

```python
len(visits)
```

```python
visits_mean = sum(visits) / len(visits)
print(f"Середнє значення: {visits_mean}")
```

```python
cov_purchases_visits = sum([(sum_purchases[i] - sum_purchases_mean) * (visits[i] - visits_mean) for i in range(0, len(visits))]) / (len(visits) - 1)
print(
    f"Вибіркова коваріація для суми покупок і кількості відвідувань: {cov_purchases_visits}")
```

```python
visits_std = (sum([(el - visits_mean) ** 2 for el in visits]) / (len(visits) - 1)) ** (1 / 2)
print(f"Стандартне відхилення: {visits_std}")
```

```python
r = cov_purchases_visits / (sum_purchases_std * visits_std)
print(
    f"Коефіцієнт кореляції Пірсона для суми покупок і кількості відвідувань (без використання бібліотек): {r}")
```

```python
r_np = np.corrcoef(sum_purchases, visits)
print(
    f"Коефіцієнт кореляції Пірсона для суми покупок і кількості відвідувань (Numpy):\n {r_np[1][0]}")
```

```python
print(f"Оскільки коефіцієнт кореляції Пірсона для суми покупок і кількості відвідувань є r = {r:.2f}, це свідчить про слабкий лінійний зв'язок.")
```
